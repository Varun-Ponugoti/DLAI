```python
import os, hashlib, random
import numpy as np, pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from PIL import Image
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, f1_score
import matplotlib.pyplot as plt

SEED = 42
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)

DATA_DIR = '/content/drive/MyDrive/IE7615_Deep_Learning_for_AI'  # <-- change this
IMG_SIZE = (224, 224)
BATCH = 32
EXTS = ('.jpg', '.jpeg')
```

```python
classes = sorted([d for d in os.listdir(DATA_DIR) if os.path.isdir(os.path.join(DATA_DIR, d))])
ll=len(classes)
print(len(classes), "folders")
for c in classes[:ll]:
    print(c, len(os.listdir(os.path.join(DATA_DIR, c))))
```

```python
rows, bad = [], []
for c in classes:
    folder = os.path.join(DATA_DIR, c)
    for f in sorted(os.listdir(folder)):
        p = os.path.join(folder, f)
        if not f.lower().endswith(EXTS):
            bad.append((p, 'not jpg/jpeg')); continue
        try:
            with Image.open(p) as im:
                im.verify()
            with Image.open(p) as im:
                rows.append((p, c, im.size, im.mode))
        except Exception as e:
            bad.append((p, f'corrupted: {e}'))

df = pd.DataFrame(rows, columns=['path', 'label', 'size', 'mode'])
print("Valid:", len(df), "| Excluded:", len(bad))
print(df['size'].value_counts().head())
print(df['mode'].value_counts())
```

```python
def md5(p):
    h = hashlib.md5()
    with open(p, 'rb') as f:
        for chunk in iter(lambda: f.read(8192), b''):
            h.update(chunk)
    return h.hexdigest()

df['hash'] = df['path'].apply(md5)
dups = df[df.duplicated('hash', keep='first')]
print("Exact duplicates removed:", len(dups))
dups[['path', 'label']].to_csv('/content/removed_duplicates.csv', index=False)  # for the report

df = df.drop_duplicates('hash', keep='first').reset_index(drop=True)
print("Final images:", len(df))
print(df['label'].value_counts().describe())
```

```python
# Duplicates: pair each removed file with the copy that was kept
kept = df[['hash', 'path', 'label']].rename(columns={'path': 'kept_path', 'label': 'kept_label'})
excluded = dups[['path', 'label', 'hash']].merge(kept, on='hash', how='left')
excluded = excluded.rename(columns={'path': 'removed_path', 'label': 'removed_label'})
excluded['same_folder'] = excluded['removed_label'] == excluded['kept_label']
excluded['reason'] = 'exact duplicate (MD5)'

print("Exact duplicates removed:", len(excluded))
print("Duplicates found across different object folders:", (~excluded['same_folder']).sum())
display(excluded[['removed_path', 'kept_path', 'same_folder']])

# Invalid / unreadable files from the validation step
bad_df = pd.DataFrame(bad, columns=['path', 'reason'])
print("Invalid files excluded:", len(bad_df))
display(bad_df)

# Save one combined log for the report
log = pd.concat([
    excluded[['removed_path', 'reason']].rename(columns={'removed_path': 'path'}),
    bad_df
])
log.to_csv('/content/excluded_images_log.csv', index=False)
print("Total excluded:", len(log))
```

```python
n = min(5, len(excluded))
if n:
    fig, axes = plt.subplots(n, 2, figsize=(6, 3*n))
    axes = np.atleast_2d(axes)
    for i in range(n):
        axes[i, 0].imshow(Image.open(excluded.loc[i, 'removed_path'])); axes[i, 0].set_title('removed', fontsize=8)
        axes[i, 1].imshow(Image.open(excluded.loc[i, 'kept_path'])); axes[i, 1].set_title('kept', fontsize=8)
        for a in axes[i]: a.axis('off')
    plt.tight_layout(); plt.show()
```

```python
classes = sorted(df['label'].unique())
class_to_idx = {c: i for i, c in enumerate(classes)}
df['y'] = df['label'].map(class_to_idx)

train_df, temp_df = train_test_split(df, test_size=0.30, stratify=df['y'], random_state=SEED)
val_df, test_df = train_test_split(temp_df, test_size=0.50, stratify=temp_df['y'], random_state=SEED)
print(len(train_df), len(val_df), len(test_df))
```

```python
AUTOTUNE = tf.data.AUTOTUNE

def load(path, y):
    img = tf.io.read_file(path)
    img = tf.image.decode_jpeg(img, channels=3)      # forces RGB
    img = tf.image.resize(img, IMG_SIZE)             # handles any dimension
    return img, y

def make_ds(d, shuffle=False):
    ds = tf.data.Dataset.from_tensor_slices((d['path'].values, d['y'].values))
    if shuffle: ds = ds.shuffle(len(d), seed=SEED)
    return ds.map(load, num_parallel_calls=AUTOTUNE).batch(BATCH).prefetch(AUTOTUNE)

train_ds = make_ds(train_df, True)
val_ds = make_ds(val_df)
test_ds = make_ds(test_df)
```

```python
augment = keras.Sequential([
    layers.RandomFlip('horizontal'),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
    layers.RandomContrast(0.1),
], name='augment')
```

```python
from tensorflow.keras.applications import MobileNetV2, ResNet50, EfficientNetB0
from tensorflow.keras.applications import mobilenet_v2, resnet50

NUM_CLASSES = len(classes)

def build_baseline():
    inp = keras.Input((224, 224, 3))
    x = augment(inp)
    x = layers.Rescaling(1./255)(x)
    for f in [32, 64, 128, 256]:
        x = layers.Conv2D(f, 3, padding='same', activation='relu')(x)
        x = layers.BatchNormalization()(x)
        x = layers.MaxPooling2D()(x)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.4)(x)
    out = layers.Dense(NUM_CLASSES, activation='softmax')(x)
    return keras.Model(inp, out, name='BaselineCNN'), None

def build_transfer(name):
    inp = keras.Input((224, 224, 3))
    x = augment(inp)
    if name == 'MobileNetV2':
        x = mobilenet_v2.preprocess_input(x)
        base = MobileNetV2(include_top=False, weights='imagenet', input_shape=(224,224,3), pooling='avg')
    elif name == 'ResNet50':
        x = resnet50.preprocess_input(x)
        base = ResNet50(include_top=False, weights='imagenet', input_shape=(224,224,3), pooling='avg')
    elif name == 'EfficientNetB0':   # has built-in normalization, expects 0-255
        base = EfficientNetB0(include_top=False, weights='imagenet', input_shape=(224,224,3), pooling='avg')
    base.trainable = False
    x = base(x, training=False)
    x = layers.Dropout(0.3)(x)
    out = layers.Dense(NUM_CLASSES, activation='softmax')(x)
    return keras.Model(inp, out, name=name), base
```

```python
def train(name, epochs_head=15, epochs_ft=15):
    model, base = build_baseline() if name == 'BaselineCNN' else build_transfer(name)
    cb = [keras.callbacks.EarlyStopping(patience=5, restore_best_weights=True, monitor='val_accuracy'),
          keras.callbacks.ReduceLROnPlateau(patience=2, factor=0.3)]
    model.compile(optimizer=keras.optimizers.Adam(1e-3 if base else 1e-3),
                  loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    h1 = model.fit(train_ds, validation_data=val_ds,
                   epochs=epochs_head if base else 40, callbacks=cb)
    hist = h1.history
    if base is not None:   # fine-tune last 30 layers
        base.trainable = True
        for l in base.layers[:-30]: l.trainable = False
        for l in base.layers:
            if isinstance(l, layers.BatchNormalization): l.trainable = False
        model.compile(optimizer=keras.optimizers.Adam(1e-5),
                      loss='sparse_categorical_crossentropy', metrics=['accuracy'])
        h2 = model.fit(train_ds, validation_data=val_ds, epochs=epochs_ft, callbacks=cb)
        for k in hist: hist[k] += h2.history[k]
    return model, hist
```

```python
results, models, histories = {}, {}, {}

def evaluate(name, model):
    y_true = np.concatenate([y.numpy() for _, y in test_ds])
    y_pred = np.argmax(model.predict(test_ds), axis=1)
    acc = (y_true == y_pred).mean()
    f1 = f1_score(y_true, y_pred, average='macro')
    results[name] = {'test_acc': acc, 'macro_f1': f1, 'params': model.count_params()}
    print(f"{name}: acc={acc:.4f}  macro-F1={f1:.4f}")
    return y_true, y_pred
```

```python
for name in ['BaselineCNN', 'MobileNetV2', 'ResNet50', 'EfficientNetB0']:
    print(f"\n===== {name} =====")
    m, h = train(name)
    models[name], histories[name] = m, h
    evaluate(name, m)
    m.save(f'/content/drive/MyDrive/{name}.keras')
```

```python
res_df = pd.DataFrame(results).T.sort_values('test_acc', ascending=False)
display(res_df)
best = res_df.index[0]
print("Best model:", best)
```

```python
plt.figure(figsize=(12, 4))
for name, h in histories.items():
    plt.plot(h['val_accuracy'], label=name)
plt.title('Validation accuracy'); plt.xlabel('epoch'); plt.legend(); plt.show()
```

```python
y_true, y_pred = evaluate(best, models[best])
print(classification_report(y_true, y_pred, target_names=classes, zero_division=0))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(14, 12)); plt.imshow(cm, cmap='Blues')
plt.title(f'{best} confusion matrix'); plt.colorbar(); plt.show()
```
